# 02 - Data Cleaning

## Objective

The objective of this notebook is to clean the raw stock market datasets before performing exploratory data analysis.

Cleaning Steps:

- Load raw datasets
- Convert Date column
- Check missing values
- Remove missing values
- Remove duplicate rows
- Sort data by Date
- Save cleaned datasets

```python
import os
import sys
from pathlib import Path

import pandas as pd 
    
sys.path.append(os.path.abspath("../src"))

%load_ext autoreload
%autoreload 2

from clean_data import data_cleaning
```

```python
raw_path = Path("../data/raw")
cleaned_path = Path("../data/cleaned")

cleaned_path.mkdir(parents=True, exist_ok=True)
```

```python
stocks = [
    "AAPL",
    "MSFT",
    "NVDA",
    "AMZN",
    "RELIANCE.NS",
    "TCS.NS",
    "HDFCBANK.NS",
    "INFY.NS"
]
```

```python
cleaned_data = {}

for stock in stocks:

    input_file = raw_path / f"{stock}.csv"
    output_file = cleaned_path / f"{stock}.csv"

    cleaned_data[stock] = data_cleaning(
        input_file,
        output_file
    )

    print(
        f"{stock}: "
        f"{len(cleaned_data[stock])} rows cleaned and saved."
    )
```

## Verify Cleaned Dataset

```python
for stock in stocks:

    df = cleaned_data[stock]

    print(
        f"{stock}: "
        f"{df['Date'].min().date()} → "
        f"{df['Date'].max().date()} | "
        f"{len(df)} rows"
    )
```

```python
for stock in stocks:

    df = cleaned_data[stock]

    print(f"\n{stock}")
    print("Missing values:", df.isnull().sum().sum())
    print("Duplicate rows:", df.duplicated().sum())
    print("Date dtype:", df["Date"].dtype)
```
