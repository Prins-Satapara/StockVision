- TCS (TCS.NS)
- HDFC Bank (HDFCBANK.NS)
- Infosys (INFY.NS)

### Data Source

- Source: Yahoo Finance
- Python Library: `yfinance`

### Data Collection Strategy

The project initially collects historical data starting from 2015-01-01.

For subsequent runs, the system checks the latest date available in each existing CSV file and fetches only the newly available market data.

The updated datasets are stored in the `data/raw` directory.

```python
import os
import sys
import pandas as pd
from pathlib import Path

sys.path.append(os.path.abspath("../src"))

%load_ext autoreload
%autoreload 2

from download_data import download_stocks
```

```python
raw_path = Path("../data/raw")

raw_path.mkdir(
    parents=True,
    exist_ok=True
)

print("Raw data directory:", raw_path)
```

```python
download_stocks(raw_path)
```

## Verify Updated Raw Datasets

```python
stocks = [
    "AAPL",
    "MSFT",
    "NVDA",
    "AMZN",
    "RELIANCE.NS",
    "TCS.NS",
    "HDFCBANK.NS",
    "INFY.NS"
]

for stock in stocks:

    file_path = raw_path / f"{stock}.csv"

    df = pd.read_csv(
        file_path,
        parse_dates=["Date"]
    )

    print(
        f"{stock}: "
        f"{df['Date'].min().date()} → "
        f"{df['Date'].max().date()} | "
        f"{len(df)} rows"
    )
```

## Observations

- Historical stock market data for eight US and Indian companies was collected from Yahoo Finance using the `yfinance` library.
- The initial datasets contain historical data starting from 2015-01-01.
- The data collection pipeline automatically checks existing CSV files and fetches newly available market data on subsequent runs.
- Only newly available data is appended to the existing raw datasets.
- The raw datasets are stored in the `data/raw` directory.
- The collected data will be used in the subsequent data cleaning and exploratory data analysis stages.
